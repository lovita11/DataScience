```python
# Q1 Berapa jumlah baris dan kolom pada dataset Titanic? Kolom apa saja yang tersedia?
import pandas as pd
import numpy as np

# Muat Dataset Titanic
URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)

print(df.shape, df.columns.tolist())
```

Pada dataset Titanic terdapat 891 baris dan 12 kolom.

Kolom-kolom yang ada pada dataset Titanic adalah :

1. PassengerId
2. Survived
3. Pclass
4. Name
5. Sex
6. Age
7. SibSp
8. Parch
9. Ticket
10. Fare
11. Cabin
12. Embarked

```python
# Q2 Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?

print(df.isnull().sum())

print(df.isnull().sum()/df.shape[0]*100)
```

Kolom yang memiliki missing value:
1. Age =  19.865320 %
2. Cabin = 77.104377 %
3. Embarked = 0.224467 %


```python
# Q3 Berapa persen penumpang yang selamat (Survived=1)?
df['Survived'].head()
jumlah_selamat = df['Survived'].sum()
total_penumpang = len(df)
presentase_selamat = (jumlah_selamat / total_penumpang) * 100
print(jumlah_selamat)
print(presentase_selamat)
```

Jumlah penumpang selamat : 342 orang

Persentase penumpang selamat : 38.38383838383838%

```python
# Q4 Berapa jumlahnya dan berapa rata- rata usianya?

wanita_kelas_1 = df[
    (df['Sex'] == 'female') &
     (df['Pclass'] == 1) &
    (df['Survived'] == 1)]
({"jumlah penumpang": [len (wanita_kelas_1)],
  "rata-rata usia": [wanita_kelas_1['Age'].mean()]
  } )
```

```python
# Q5 Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?

survival_rate = df.groupby('Pclass')['Survived'].mean() * 100
kelas_tertinggi = survival_rate.idxmax()
tingkat_keselamatan_tertinggi = survival_rate.max()
print(survival_rate)
```

Kelas penumpang yan memiliki keselamatan tertinggi = kelas 1

```python
# Q6 Buat analisis tambahan berdasarkan variabel lain yang Anda anggap menarik. Sertakan interpretasi singkat.
hasil = df.groupby('Sex').agg(
  total=('PassengerId','count'),
  selamat=('Survived','mean'))

print(hasil)

```
